### Paso 4: Previsualizar cada archivo `.dbf`

Ahora vamos a recorrer todos los archivos `.dbf` encontrados en la carpeta y mostraremos las primeras filas de cada uno utilizando `dbfread` y `pandas`.

```python
import pandas as pd
from dbfread import DBF
import os

# Define la ruta de la carpeta donde están los archivos DBF
folder_path = '/content/drive/My Drive/Riegos del sur/Datos Memory'

# Lista todos los archivos .dbf en la carpeta
dbf_files = [f for f in os.listdir(folder_path) if f.endswith('.DBF') or f.endswith('.dbf')]

if not dbf_files:
    print(f"No se encontraron archivos DBF en la carpeta: {folder_path}")
else:
    print(f"Se encontraron {len(dbf_files)} archivos DBF. Previsualizando cada uno:")
    for dbf_file_name in dbf_files:
        dbf_file_path = os.path.join(folder_path, dbf_file_name)
        try:
            # Leer el archivo DBF
            # Se usa encoding='latin-1' o 'cp850' que son comunes para archivos DBF
            table = DBF(dbf_file_path, encoding='latin-1')
            df_dbf = pd.DataFrame(iter(table))

            print(f"\n--- Previsualización de '{dbf_file_name}' ---")
            display(df_dbf.head())
            print(f"Número de filas: {len(df_dbf)}")
            print(f"Columnas y tipos de datos:\n{df_dbf.info(verbose=False, memory_usage='deep')}")

        except Exception as e:
            print(f"Ocurrió un error al cargar '{dbf_file_name}': {e}")
```

### Análisis del archivo `ARDATOS.DBF`

```python
# Especifica el nombre del archivo DBF a analizar
ardatos_file_name = 'ARDATOS.DBF'
ardatos_file_path = os.path.join(folder_path, ardatos_file_name)

try:
    # Cargar el archivo ARDATOS.DBF en un DataFrame de pandas
    print(f"Cargando el archivo: {ardatos_file_name}")
    table_ardatos = DBF(ardatos_file_path, encoding='latin-1')
    df_ardatos = pd.DataFrame(iter(table_ardatos))

    print(f"\n--- Primeras 5 filas de '{ardatos_file_name}' ---")
    display(df_ardatos.head())

    print(f"\n--- Información general de '{ardatos_file_name}' ---")
    df_ardatos.info(verbose=True, memory_usage='deep')

    print(f"\n--- Estadísticas descriptivas de '{ardatos_file_name}' ---")
    display(df_ardatos.describe(include='all'))

    print(f"\n--- Conteo de valores nulos por columna en '{ardatos_file_name}' ---")
    display(df_ardatos.isnull().sum().to_frame(name='valores_nulos'))

except Exception as e:
    print(f"Ocurrió un error al cargar o analizar '{ardatos_file_name}': {e}")
```
